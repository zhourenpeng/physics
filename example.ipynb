```python
#定义函数设置参数
def setparameter(u_nudm):
    LambdaCDM.empty()
    LambdaCDM.set({
        'omega_b':omega_b,  
        'omega_cdm':omega_cdm,  
        'H0':H0,  
        'A_s':A_s, 
        'n_s':n_s,  
        'tau_reio':tau_reio,
        'N_ur':0.00641,
        'N_ncdm':1,
        'deg_ncdm':3,
        'm_ncdm':0.02,
        'T_ncdm':T_ncdm,
        'YHe':YHe,
        'u_ncdmdm':u_nudm
    })
    LambdaCDM.set({'output':'mPk,lCl,tCl,pCl,vTk,dTk',
               'l_max_scalars':3000,
               'headers':'yes',
               'format':'class',
               'z_max_pk':0,
               # 'non linear':'hmcode',
               'write_exotic_injection':'yes',
               'write_distortions':'yes',
               'write_background':'yes',
               'write thermodynamics':'yes',
               'write_primordial':'yes',
               'lensing':'yes',
               'P_k_max_1/Mpc':10,
               'gauge':'newtonian',
               # 'nonlinear_min_k_max':10,
               # 'non linear':'halofit',
               'root':'/home/zhourenpeng/downloads/class_public-3.2.1/output/base_2018_plikHM_TTTEEE_lowl_lowE_lensing',
               'write warnings':'yes',
               'write_parameters':'yes',
               'input_verbose':1,
               'background_verbose':0,
               'thermodynamics_verbose':0,
               'perturbations_verbose':0,
               'transfer_verbose':0,
               'primordial_verbose':0,'harmonic_verbose':0,
               'fourier_verbose':0,
               'lensing_verbose':0,
               'output_verbose':0
              })
    #LambdaCDM.set({'k_output_values': '0.01,0.1,0.001'})
    LambdaCDM.compute()
    return LambdaCDM
```
